```python
#uniform() - for all rows 
IMDB=np.random.uniform(low=0,high=10,size=len(netflix['title']))
#to round up decimal - around()
IMDB=np.around(IMDB,decimals=1)
netflix['IMDB Rating']=IMDB
netflix
```

Find how many items have 10 IMDB Rating (logical masking)

```python
netflix[netflix['IMDB Rating']==10].count()
```

```python
netflix[netflix['IMDB Rating']==9.6]['title']
```

```python
netflix.sort_values(by='IMDB Rating',ascending=False)
```

```python
#Get the items with IMDB rating bigger than 8.0 and country with spain
netflix[(netflix['IMDB Rating']>8) & (netflix['country']=='Spain')]
```

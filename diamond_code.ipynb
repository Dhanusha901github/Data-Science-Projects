Create a column called price_per_carat

```python
price_per_carat=(diamond['price']/diamond['carat']).round(0)
diamond['price_per_carat']=price_per_carat
diamond
```

Find the diamond with the highest price_per_carat

```python
diamond.sort_values('price_per_carat',ascending=False).head(1)
```

Create a new column:
Small (carat < 1)
Medium (1–2)
Large (>2)

```python
small=diamond['carat']<1
medium=diamond['carat'].between(1,2)
large=diamond['carat']>2
diamond['small']=small
diamond['medium']=medium
diamond['large']=large
diamond
```

Count diamonds in each size category

```python
diamond[['small','medium', 'large']].sum()
```

Find the average price for each size category

```python
diamond[diamond['small']]['price'].mean()
```

```python
diamond[diamond['medium']]['price'].mean()
```

```python
diamond[diamond['large']]['price'].mean()
```

Scatter plot of carat vs price

```python
plt.scatter(diamond['carat'], diamond['price'], color='green', marker='o',s=0.7)
plt.xlabel('Carat', fontsize=10, family='Times New Roman')
plt.ylabel('Price', fontsize=10, family='Times New Roman')
plt.title('Carat vs Price', fontsize=15,family='Arial')
plt.show()
```

Scatter plot of depth vs price

```python
plt.scatter(diamond['depth'], diamond['price'], color='cyan', marker='s',s=1.5)
plt.xlabel('Depth', fontsize=10, family='Times New Roman')
plt.ylabel('Price', fontsize=10, family='Times New Roman')
plt.title('Depth vs Price', fontsize=15,family='Arial')
plt.show()
```
